```python
import pandas as pd

# Memuat dataset dari URL
url = "https://github.com/datasciencedojo/datasets/raw/master/titanic.csv"
df = pd.read_csv(url)
```

```python
#Q1 : Berapa jumlah baris dan kolom pada dataset Titanic? Kolom apa saja yang tersedia?
print("Q1 : Berapa jumlah baris dan kolom pada dataset Titanic? Kolom apa saja yang tersedia?")
jawaban1 = df.shape
print(jawaban1)
print("Berdasarkan hasil tersebut dataset Titanic memiliki 891 baris dan 12 kolom \n\n")
```

```python
#Q2 : Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?
print("Q2 : Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?")
# Menghitung jumlah missing values per kolom
missing_values = df.isnull().sum()
print(missing_values)
# Menghitung persentase missing values per kolom
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)
print("\nDari data diatas dapat terlihat bahwa kolom yang memiliki missing values ada di kolom Age, Cabin, Embarked \nJumlah nya masing masing 177, 687, 2 \nPresentase masing masing 19,8%, 77,1% 0,2% \n\n")
```

```python
#Q3 : Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPy atau Pandas.
print("Q3 : Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPy atau Pandas")
jawaban3 = df['Survived'].mean() * 100
print(jawaban3)
print(f"Berdasarkan hasil tersebut maka Persentase penumpang yang selamat: {jawaban3:.2f}% \n\n")
```

```python
#Q4 : Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa rata- rata usianya?
print("Q4 Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa rata- rata usianya?")
# Filter penumpang wanita (female) dari kelas 1 (Pclass=1)
penumpang_wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
#hitung jumlahnya
jawaban4a = len(penumpang_wanita_kelas1)
print(jawaban4a)
#hitung rata rata age nya
jawaban4b = penumpang_wanita_kelas1['Age'].mean()
print(jawaban4b)
print(f"Berdasarkan hasil tersebut maka penumpang wanita dari kelas 1 berjumlah {jawaban4a} orang dan rata rata usianya = {jawaban4b:.2f}% \n\n")
```

```python
#Q5 : Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?
print("Q5 Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?")
# Mengelompokkan data berdasarkan 'Pclass' dan menghitung rata-rata 'Survived'
jawaban5 = df.groupby('Pclass')['Survived'].mean() * 100
print(jawaban5)
print("Berdasarkan hasil tersebut maka kelas yang paling tinggi tingkat keselamtannya adalah Kelas 1 \n\n")
```

```python
#Q6 : Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat.
print("Q6 : Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat")
print("Analisis yang akan saya buat adalah presentase gender terhadap tingkat keselamatan")
# Analisis tingkat keselamatan berdasarkan Gender
keselamatan_gender = df.groupby('Sex')['Survived'].mean() * 100
print(keselamatan_gender.map('{:.2f}%'.format))
print("\nDari hasil diatas maka Wanita memiliki Tingkat Keselamatan Lebih Besar dibandingkan Laki Laki")
```
