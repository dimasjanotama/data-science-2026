#1. Muat dan Eksplorasi Data
- Pada case ini menggunakan dataset Telco Churn dari Kaggle

```python
import pandas as pd

df = pd.read_csv("telco_churn.csv")
print(df.shape)
print(df["Churn"].value_counts(normalize=True))
```

#2. Preprocessing
- Encoding pada fitur kategorikal menggunakan pd.get_dummies() atau LabelEncoder
- Bagi data menjadi latih/uji secara stratified untuk menjaga proporsi kelas

```python
from sklearn.model_selection import train_test_split

# TODO: encoding fitur kategorikal (mis. pd.get_dummies)
# TODO: pisahkan X (fitur) dan y (target = Churn)
X_raw = df.drop(columns=["Churn"])
y = df["Churn"]
X = pd.get_dummies(X_raw, drop_first=True)

X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
```

#3. Latih Model
- Bangun RandomForestClassifier dengan class_weight="balanced" untuk menangani imbalance

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=300, class_weight="balanced", random_state=42)
rf.fit(X_tr, y_tr)
```

#4. Evaluasi
- Hitung precision, recall, F1-score, dan ROC-AUC dengan fokus pada kelas churn (kelas 1)

```python
from sklearn.metrics import classification_report, roc_auc_score

# TODO: hitung prediksi dan probabilitas
# Mengambil prediksi kelas (0 atau 1) untuk classification report
y_pred = rf.predict(X_te)
# Mengambil probabilitas kelas positif (Churn).
y_prob = rf.predict_proba(X_te)[:, 1]

# TODO: tampilkan classification_report dan ROC-AUC
print("================ CLASSIFICATION REPORT ================")
print(classification_report(y_te, y_pred))
print("=======================================================")

roc_auc = roc_auc_score(y_te, y_prob)
print(f"ROC-AUC Score: {roc_auc:.4f}")
```

#5. Prediksi Probabilitas dan Simpulkan
- Menggunakan predict_proba untuk menghasilkan probabilitas churn tiap pelanggan
- Kesimpulan mengenai temuan utama dari model

```python
# TODO: hitung probabilitas churn (predict_proba)
# Probabilitas untuk kelas positif (Churn)
y_prob_churn = rf.predict_proba(X_te)[:, 1]
print("Array probabilitas churn tiap pelanggan:")
print(y_prob_churn)


# TODO: tulis kesimpulan singkat (3-4 kalimat)
```

Model Random Forest yang dibangun berhasil memprediksi perilaku churn pelanggan dengan performa yang baik, ditunjukkan oleh skor ROC-AUC sebesar 0.8337. Secara keseluruhan, model ini sudah siap digunakan untuk mengidentifikasi pelanggan yang berisiko tinggi dan mulai merumuskan strategi retensi.
